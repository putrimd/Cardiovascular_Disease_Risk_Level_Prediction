```python
# Import library

import pickle
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
# load model
with open ('deployment/src/best_gradboost_model.pkl', 'rb') as model_file:
    best_gradboost_model = pickle.load(model_file)

# load label encoder
with open ('deployment/src/label_encoder.pkl', 'rb') as label_encoder_file:
    le = pickle.load(label_encoder_file)
```

```python
# input new data
data_inf = {
    'Sex': 'F',
    'Age': 30,
    'Weight': 55,
    'Height_m': 1.63,
    'BMI': 20.7,
    'Abdominal Circumference': 75.5,
    'Blood Pressure (mmHg)': '104/77',
    'Total Cholesterol': 211,
    'HDL': 78,
    'Fasting Blood Sugar': 100,
    'Smoking Status': 'N',
    'Diabetes Status': 'N',
    'Physical Activity Level': 'Moderate',
    'Family History of CVD': 'Y',
    'Height_cm' : 163,
    'Waist-to-Height Ratio': 0.46,
    'Systolic BP': 104,
    'Diastolic BP': 77,
    'Blood Pressure Category': 'Normal',
    'Estimated LDL': np.nan ,
    'CVD Risk Score': np.nan
}

# input new data to dataframe
data_inf = pd.DataFrame([data_inf])

# show new data
data_inf
```

```python
# predict target for new user
y_pred_original = best_gradboost_model.predict(data_inf)

# encode target label
y_pred = le.inverse_transform(y_pred_original)

# hasil prediksi
result = pd.DataFrame({
    'prediction': y_pred
})

# show result
result
```

### Conclusion & Recommendation

1. New user Female 30 tahun tersebut diprediksi memiliki **CVD Risk Level Intermediary**
2. Rekomendasi: Dibutuhkan konsultasi dengan dokter spesialis dan rutin untuk cek kesehatan untuk penanganan yang sesuai dengan kebutuhan. Menjaga pola makan dan aktivitas fisik agar dapat menurunkan resiko penyakit Cardiovascular.
